4. How do you pass inputs to a LangChain chain?

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptsTemplate

llm = ChatOpenAI(
    model="openai-4o-mini"
)
prompt = ChatPromptsTemplate.form_template(
    "Explain {topic} in {tone} terms."
)

chain = prompt | llm

# Commonly used.
result  = chain.invoke({
    "topic": "Machine Learning",
    "tone": "simple"
})

# old method
result  = chain.run({
    "topic": "LLM"
})

# Batch Process
result = chain.batch([
    {"topic": "AI"},
    {"topic": "LLM"},
    {"topic": "NLP"}
])

print(result.content)

#4. Streaming input

for chunk in chain.stream({"topic": "AI"}):
    print(chunk.content, end="")

```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

llm = ChatOllama(model="llama3.2:latest", temperature=0.7, max_tokens=100)

"""
prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in 50 words."
)
"""

prompt = ChatPromptTemplate.from_messages([
    ("system","You are helpful assistant."),
    MessagesPlaceholder("history"),
    ("human", "{input}")
])

base_chain = prompt | llm

# store history with session ID
_store = {}

def get_history(session_id: str):
    if session_id not in _store:
        _store[session_id] = InMemoryChatMessageHistory()
    return _store[session_id]

chat_chain = RunnableWithMessageHistory(
    base_chain,
    get_history,
    input_messages_key="input",
    history_messages_key="history"
)

# Usage:
resp1 = chat_chain.invoke(
    {"input": "My name is Nagesh."},
    config={"configurable": {"session_id": "user-1"}}
)
resp2 = chat_chain.invoke(
    {"input": "What is my name?"},
    config={"configurable": {"session_id": "user-1"}}
)
history = get_history("user-1");
print(history)
print(resp2.content)
```

7. Implement a chain to handle batch processing in LangChain.


```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

def build_batch_chain(model: str = "gpt-4o-mini"):
    llm = ChatOllama(model="llama3.2:latest", temperature=0.0, max_tokens=200)
    prompt = ChatPromptTemplate.from_template("Summarize in one sentence: {text}")
    return prompt | llm

chain = build_batch_chain()

inputs = [
    {"text": "LangChain helps build LLM apps using chains and tools."},
    {"text": "Transformers are neural networks that handle sequences well."},
    {"text": "MoE activates only a subset of experts per token."},
]

outputs = chain.batch(inputs)
for o in outputs:
    print(o.content)
```

9. Write a function to implement a multi-agent system in LangChain.

```python
from langchain.chat_models import init_chat_model
from langchain.agents import create_react_agent, AgentExecutor

def make_agent(model_name: str, tools):
    model = init_chat_model(model_name, temperature=0.2)
    agent = create_react_agent(llm=model, tools=tools)
    return AgentExecutor(agent=agent, tools=tools, verbose=True)

def multi_agent_pipeline(question: str, research_tools, writing_tools):
    researcher = make_agent("llama3.2:latest", research_tools)
    writer = make_agent("llama3.2:latest", writing_tools)

    research_notes = researcher.invoke({"input": f"Research this and return notes:\\n{question}"})["output"]
    final_answer = writer.invoke({"input": f"Using these notes, write a clean final answer:\\n\\n{research_notes}"})["output"]

    return final_answer
```
