```python
# integer
num = 10
print(num)

# float
num = 10.5
print(num)

# string
str = "Hello"
print(str)

# boolean
bool = True
print(bool)
```

Python built-in data types

**LIST**

```python
# ordered collection with duplicates.
list = [1, 2, 3, 4, 5]
print(list)
```

** tuple **

```python
# ordered collection with no duplicates.
tup = (1,2,3,4,5)
print(tup)
```

** dictionary **  

```python
# key-value pairs
dict = {"name": "Nagesh", "age": 30, "city": "Pune"}
print(dict)
```

** SET **
# Unordered collection with no duplicates.

```python
unique_id = {1,2,3,4,4,5}
print(unique_id)
```

What is the difference between == and is?

```python
# == checks for value equality

A = [1,3]
B = [1,3]

print(A == B)
```

```python
# uses memory address to compare objects
print(A is B)
```

```python
# Function is a block of code that performs a specific task
def name(name):
    print(name)
name("Nagesh")
```

```python
# What is the difference between parameters and arguments?
def add(a, b): # a and b are parameters
    return a + b

add(1, 2) # 1 and 2 are arguments

```
