```python
from fastapi import FastAPI

app = FastAPI()
@app.get("/users")
def read_users():
    return [{"username": "Alice"}, {"username": "Bob"}]
```

```python
from fastapi import Depends, FastAPI
app = FastAPI()

async def get_data():
    return [{"username": "Alice"}, {"username": "Bob"}]

@app.get("/get-users")
async def read_users(data: list = Depends(get_data)):
    print(data)
```
